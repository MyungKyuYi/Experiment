```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# 1. 기본 경로 및 설정
base_path = '/content/drive/MyDrive/Colab Notebooks/PPG_REAL'
subject_ids = range(1, 11)  # 1번부터 10번까지

# 그래프 한글 폰트 설정 (필요 시 주석 해제)
# !sudo apt-get install -y fonts-nanum
# !sudo fc-cache -fv
# !rm ~/.cache/matplotlib -rf
# plt.rc('font', family='NanumBarunGothic')

for subject_id in subject_ids:
    # 2. 파일 경로 생성
    file_cart_name = f'E{subject_id:04d}_cart.xlsx'
    file_pwv_name = f'PWV{subject_id}.xlsx'

    path_cart = os.path.join(base_path, file_cart_name)
    path_pwv = os.path.join(base_path, file_pwv_name)

    print(f"\n[{subject_id}번 대상자 처리 중...]")

    if not os.path.exists(path_cart) or not os.path.exists(path_pwv):
        print(f"  => 파일을 찾을 수 없어 건너뜁니다: {file_cart_name}, {file_pwv_name}")
        continue

    try:
        # 3. 데이터 로드
        df_cart = pd.read_excel(path_cart)
        df_pwv = pd.read_excel(path_pwv)

        # 4. 전처리
        df_cart['datetime'] = pd.to_datetime(df_cart['날짜'].astype(str) + ' ' + df_cart['시간'].astype(str))
        df_pwv['datetime'] = pd.to_datetime(df_pwv['Date'].astype(str) + ' ' + df_pwv['Time'].astype(str))

        df_cart = df_cart.rename(columns={'수축기 (SBP)': 'SBP', '이완기 (DBP)': 'DBP', '중간혈압 (MBP)': 'MAP', '수면 (SLEEP)': 'SLEEP'})
        df_pwv = df_pwv.rename(columns={'Systole': 'SBP', 'Diastole': 'DBP', 'MAP': 'MAP'})

        # 5. 교집합 구간 설정
        start_time = max(df_cart['datetime'].min(), df_pwv['datetime'].min())
        end_time = min(df_cart['datetime'].max(), df_pwv['datetime'].max())

        if start_time > end_time:
            print(f"  => 겹치는 시간 구간 없음")
            continue

        cart_plot = df_cart[(df_cart['datetime'] >= start_time) & (df_cart['datetime'] <= end_time)].sort_values('datetime')
        pwv_plot = df_pwv[(df_pwv['datetime'] >= start_time) & (df_pwv['datetime'] <= end_time)].sort_values('datetime')

        if cart_plot.empty or pwv_plot.empty:
            print(f"  => 필터링 후 데이터 없음")
            continue

        # 6. 수면 구간 추출
        sleep_intervals = []
        in_sleep = False
        sleep_start = None
        for i, row in cart_plot.iterrows():
            if row['SLEEP'] == 1:
                if not in_sleep:
                    in_sleep = True
                    sleep_start = row['datetime']
            else:
                if in_sleep:
                    in_sleep = False
                    sleep_intervals.append((sleep_start, row['datetime']))
                    sleep_start = None
        if in_sleep:
            sleep_intervals.append((sleep_start, cart_plot.iloc[-1]['datetime']))

        # 7. 그래프 그리기
        fig, axes = plt.subplots(3, 1, figsize=(12, 12), sharex=True)
        metrics = ['SBP', 'DBP', 'MAP']
        colors = {'Cart': 'blue', 'PWV': 'red'}

        for i, metric in enumerate(metrics):
            ax = axes[i]
            ax.plot(cart_plot['datetime'], cart_plot[metric], label='Cart', marker='o', color=colors['Cart'], alpha=0.7)
            ax.plot(pwv_plot['datetime'], pwv_plot[metric], label='PWV', marker='x', color=colors['PWV'], alpha=0.7)

            # 수면 배경
            for s_start, s_end in sleep_intervals:
                if s_start == s_end:
                    ax.axvline(s_start, color='gray', alpha=0.3, lw=5)
                else:
                    ax.axvspan(s_start, s_end, color='gray', alpha=0.3)

            ax.set_ylabel(f'{metric} (mmHg)')
            ax.legend(loc='upper right')
            ax.grid(True, linestyle='--', alpha=0.5)
            ax.set_title(f'Subject {subject_id} - {metric}')

        axes[-1].set_xlabel('Time')
        axes[-1].xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
        plt.xticks(rotation=45)
        plt.tight_layout()

        # 저장 및 출력
        save_path = os.path.join(base_path, f'comparison_plot_{subject_id:02d}.png')
        plt.savefig(save_path)
        print(f"  => 저장 완료: {save_path}")

        plt.show()  # <--- 이 부분이 추가되었습니다. (화면에 출력)
        plt.close() # 메모리 관리를 위해 닫기

    except Exception as e:
        print(f"  => 에러 발생: {e}")

print("\n모든 작업 완료.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
import numpy as np

# ==========================================
# 1. 기본 설정 (경로를 본인 환경에 맞게 수정하세요)
# ==========================================
base_path = '/content/drive/MyDrive/Colab Notebooks/PPG_REAL'
subject_ids = range(1, 11)  # 1번부터 10번까지

# 한글 폰트 설정 (Colab용, 필요시 주석 해제)
# !sudo apt-get install -y fonts-nanum
# !sudo fc-cache -fv
# !rm ~/.cache/matplotlib -rf
# plt.rc('font', family='NanumBarunGothic')

# ==========================================
# 2. 핵심 함수 정의
# ==========================================
def calculate_mae(y_true, y_pred):
    """평균 절대 오차(MAE) 계산"""
    return np.mean(np.abs(y_true - y_pred))

def remove_outliers(df, col_sbp='SBP', col_dbp='DBP'):
    """생리학적으로 비현실적인 혈압 데이터 제거 (Physiological Filter)"""
    # SBP: 80 ~ 200 mmHg / DBP: 40 ~ 130 mmHg 범위를 벗어나면 이상치로 간주
    condition = (df[col_sbp] >= 80) & (df[col_sbp] <= 210) & \
                (df[col_dbp] >= 40) & (df[col_dbp] <= 130)
    return df[condition]

# ==========================================
# 3. 메인 루프
# ==========================================
for subject_id in subject_ids:
    file_cart_name = f'E{subject_id:04d}_cart.xlsx'
    file_pwv_name = f'PWV{subject_id}.xlsx'

    path_cart = os.path.join(base_path, file_cart_name)
    path_pwv = os.path.join(base_path, file_pwv_name)

    print(f"\n[{subject_id}번 대상자 분석 시작...]")

    # 파일 존재 여부 확인
    if not os.path.exists(path_cart) or not os.path.exists(path_pwv):
        print(f"  => 파일을 찾을 수 없어 건너뜁니다.")
        continue

    try:
        # 데이터 로드
        df_cart = pd.read_excel(path_cart)
        df_pwv = pd.read_excel(path_pwv)

        # 날짜/시간 컬럼 합치기 및 변환
        df_cart['datetime'] = pd.to_datetime(df_cart['날짜'].astype(str) + ' ' + df_cart['시간'].astype(str))
        df_pwv['datetime'] = pd.to_datetime(df_pwv['Date'].astype(str) + ' ' + df_pwv['Time'].astype(str))

        # 컬럼명 통일
        df_cart = df_cart.rename(columns={'수축기 (SBP)': 'SBP', '이완기 (DBP)': 'DBP', '중간혈압 (MBP)': 'MAP', '수면 (SLEEP)': 'SLEEP'})
        df_pwv = df_pwv.rename(columns={'Systole': 'SBP', 'Diastole': 'DBP', 'MAP': 'MAP'})

        # [전략 1] Reference(PWV) 이상치 제거
        original_count = len(df_pwv)
        df_pwv = remove_outliers(df_pwv)
        dropped_count = original_count - len(df_pwv)
        if dropped_count > 0:
            print(f"  => PWV 이상치(노이즈) {dropped_count}개 제거됨")

        # 시간 교집합 구간 설정
        start_time = max(df_cart['datetime'].min(), df_pwv['datetime'].min())
        end_time = min(df_cart['datetime'].max(), df_pwv['datetime'].max())

        # 데이터 필터링
        cart_plot = df_cart[(df_cart['datetime'] >= start_time) & (df_cart['datetime'] <= end_time)].sort_values('datetime')
        pwv_plot = df_pwv[(df_pwv['datetime'] >= start_time) & (df_pwv['datetime'] <= end_time)].sort_values('datetime')

        if cart_plot.empty or pwv_plot.empty:
            print("  => 유효한 시간 구간 내 데이터가 없습니다.")
            continue

        # [전략 2] 데이터 매칭 (Time Synchronization)
        # tolerance='10min': 10분 이내의 가장 가까운 데이터를 매칭
        merged_df = pd.merge_asof(
            cart_plot.sort_values('datetime'),
            pwv_plot.sort_values('datetime'),
            on='datetime',
            direction='nearest',
            tolerance=pd.Timedelta('10min'),
            suffixes=('_cart', '_pwv')
        ).dropna()

        if merged_df.empty:
            print("  => 시간 동기화 실패 (매칭되는 데이터 없음)")
            continue

        # [전략 3] 캘리브레이션 및 지표 계산
        metrics = ['SBP', 'DBP', 'MAP']
        offsets = {}
        mae_results = {}

        for metric in metrics:
            # 보정값(Offset) 계산: (Reference 평균 - SmartRing 평균)
            offset = merged_df[f'{metric}_pwv'].mean() - merged_df[f'{metric}_cart'].mean()
            offsets[metric] = offset

            # 그래프용 데이터프레임에 보정된 컬럼 추가
            cart_plot[f'{metric}_cal'] = cart_plot[metric] + offset

            # MAE 계산 (Raw vs Calibrated)
            mae_raw = calculate_mae(merged_df[f'{metric}_pwv'], merged_df[f'{metric}_cart'])
            mae_cal = calculate_mae(merged_df[f'{metric}_pwv'], merged_df[f'{metric}_cart'] + offset)
            mae_results[metric] = (mae_raw, mae_cal)

        # 수면 구간 추출 (그래프 배경용)
        sleep_intervals = []
        in_sleep = False
        sleep_start = None
        for i, row in cart_plot.iterrows():
            if row['SLEEP'] == 1:
                if not in_sleep:
                    in_sleep = True
                    sleep_start = row['datetime']
            else:
                if in_sleep:
                    in_sleep = False
                    sleep_intervals.append((sleep_start, row['datetime']))
                    sleep_start = None
        if in_sleep:
            sleep_intervals.append((sleep_start, cart_plot.iloc[-1]['datetime']))

        # ==========================================
        # 4. 시각화
        # ==========================================
        fig, axes = plt.subplots(3, 1, figsize=(14, 12), sharex=True)
        colors = {'Cart_Raw': 'blue', 'Cart_Cal': 'green', 'PWV': 'red'}

        for i, metric in enumerate(metrics):
            ax = axes[i]

            # 1. Raw Cart Data (파란 점선)
            ax.plot(cart_plot['datetime'], cart_plot[metric],
                    label='Cart (Raw)', color=colors['Cart_Raw'], linestyle=':', alpha=0.5)

            # 2. Calibrated Cart Data (초록 실선 - 핵심!)
            ax.plot(cart_plot['datetime'], cart_plot[f'{metric}_cal'],
                    label='Cart (Calibrated)', color=colors['Cart_Cal'], linewidth=2, alpha=0.9)

            # 3. Reference PWV (빨간 엑스)
            ax.plot(pwv_plot['datetime'], pwv_plot[metric],
                    label='Reference (PWV)', marker='x', color=colors['PWV'], linestyle='-', alpha=0.6)

            # 수면 구간 표시
            for s_start, s_end in sleep_intervals:
                if s_start == s_end:
                    ax.axvline(s_start, color='gray', alpha=0.2, lw=5)
                else:
                    ax.axvspan(s_start, s_end, color='gray', alpha=0.2)

            # 타이틀 설정 (MAE 변화 표시)
            mae_pre, mae_post = mae_results[metric]
            offset_val = offsets[metric]
            ax.set_title(f'Subject {subject_id} - {metric} | MAE: {mae_pre:.1f} → {mae_post:.1f} mmHg (Offset: {offset_val:.1f})',
                         fontsize=12, fontweight='bold')
            ax.set_ylabel(f'{metric} (mmHg)')
            ax.grid(True, linestyle='--', alpha=0.5)

            if i == 0:
                ax.legend(loc='upper right')

        axes[-1].set_xlabel('Time')
        axes[-1].xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
        plt.xticks(rotation=45)
        plt.tight_layout()

        # 이미지 저장
        save_path = os.path.join(base_path, f'Result_Subj{subject_id}_Calibrated.png')
        plt.savefig(save_path)
        print(f"  => 그래프 저장 완료: {save_path}")

        plt.show()
        plt.close()

    except Exception as e:
        print(f"  => 에러 발생: {e}")

print("\n모든 분석이 완료되었습니다.")
```
